```
#import all important library
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz
```

```
#Collecting Dataset
df = pd.read_csv('/content/Play Store Data.csv')
```

```
#Missing Value Handling
df.isnull().sum()
```

```
df.dropna()
```

```
#Duplicate Handling
df.duplicated().sum()
```

```
df.drop_duplicates()
```

```
#Dealing With Data Inconsistency
df['Installs'] = (
    df['Installs']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False)
)
```

```
df = df[df['Installs'].str.isnumeric()]
```

```
df['Installs'] = df['Installs'].astype(int)
```

```
excluded_letters = ('A', 'C', 'G', 'S')

filtered_df = df[
    ~df['Category'].str.startswith(excluded_letters, na=False)
]
```

```
category_installs = (
    filtered_df
    .groupby('Category', as_index=False)['Installs']
    .sum()
)
```

```
top5 = (
    category_installs
    .sort_values(by='Installs', ascending=False)
    .head(5)
)
```

```
top5['Highlight'] = top5['Installs'].apply(
    lambda x: 'Above 1M' if x > 1_000_000 else 'Below 1M'
)
```

```
sample_countries = [
    'United States',
    'India',
    'Brazil',
    'Germany',
    'Canada'
]
```

```
top5['Country'] = sample_countries[:len(top5)]
```

```
#Creating Map
ist = pytz.timezone('Asia/Kolkata')
current_time = datetime.now(ist)
current_hour = current_time.hour

if 18 <= current_hour < 20:

    # =========================
    # CREATE CHOROPLETH MAP
    # =========================

    fig = px.choropleth(
        top5,
        locations='Country',
        locationmode='country names',
        color='Installs',
        hover_name='Category',
        hover_data=['Installs', 'Highlight'],
        title='Global Installs by Top 5 App Categories',
        color_continuous_scale='Viridis'
    )
    fig.update_layout(
        template='plotly_dark',
        height=600,
        title_x=0.5
    )

    fig.show()
else:
    print(
        'Dashboard hidden. Graph is available only between 6 PM IST and 8 PM IST.'
    )
```
